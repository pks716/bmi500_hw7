# HW1 Part B - SIR Model Implementation with Values

SIR equations:

- dS/dt = -beta * S * I
- dI/dt = beta * S * I - gamma * I
- dR/dt = gamma * I

I am using the forward Euler method to solve it:

new value = old value + dt * (derivative)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sir_model(S0, I0, R0, beta, gamma, days, dt=0.1):
    # number of steps
    n = int(days / dt)

    t = np.zeros(n + 1)
    S = np.zeros(n + 1)
    I = np.zeros(n + 1)
    R = np.zeros(n + 1)

    # starting values
    S[0] = S0
    I[0] = I0
    R[0] = R0

    for k in range(n):
        dS = -beta * S[k] * I[k]
        dI = beta * S[k] * I[k] - gamma * I[k]
        dR = gamma * I[k]

        S[k + 1] = S[k] + dt * dS
        I[k + 1] = I[k] + dt * dI
        R[k + 1] = R[k] + dt * dR
        t[k + 1] = t[k] + dt

    return t, S, I, R
```

### Simulation with Part B values


```python
t, S, I, R = sir_model(1995, 5, 0, 0.125e-3, 0.1, 160)

plt.plot(t, S, label='Susceptible (S)')
plt.plot(t, I, label='Infected (I)')
plt.plot(t, R, label='Recovered (R)')
plt.xlabel('Days')
plt.ylabel('Number of people')
plt.title('SIR model, N = 2000, 160 days')
plt.legend()
plt.show()
```
